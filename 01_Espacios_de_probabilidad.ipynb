## 1. Espacio de Probabilidad $(\Omega, \mathcal{F}, P)$

La teoría axiomática moderna de la probabilidad (formulada por A. N. Kolmogorov en 1933) fundamenta el estudio del azar sobre la **Teoría de la Medida**. Un **Espacio de Probabilidad** es una terna ordenada $(\Omega, \mathcal{F}, P)$ definida sobre un experimento aleatorio $\mathcal{E}$:

Definiendo lo siguiente:

### 1. Espacio Muestral ($\Omega$)
Es un conjunto no vacío ($\Omega \neq \emptyset$) que contiene a la totalidad de los resultados elementales posibles $\omega$ del experimento $\mathcal{E}$, tal que $\Omega = \{\omega : \omega \text{ es un resultado posible}\}$. 
* Puede ser de naturaleza **discreta** (finito o contable) o **continua** (no contable, p. ej., subconjuntos de $\mathbb{R}^n$).

---

### 2. $\sigma$-álgebra ($\mathcal{F}$)
Sea $\mathcal{P}(\Omega)$ el conjunto potencia de $\Omega$. Una **$\sigma$-álgebra** (o $\sigma$-campo) $\mathcal{F}$ es una familia no vacía de subconjuntos de $\Omega$ ($\mathcal{F} \subseteq \mathcal{P}(\Omega)$) que constituye la estructura medible del espacio. Cada elemento $A \in \mathcal{F}$ recibe el nombre de **evento u objeto medible**.

$\mathcal{F}$ satisface estrictamente las siguientes tres condiciones algebraicas:

1. **Pertenencia del Espacio Total:**
   $$\Omega \in \mathcal{F}$$
2. **Cierre bajo Complementación:**
   $$\text{Si } A \in \mathcal{F}, \implies A^c = \Omega \setminus A \in \mathcal{F}$$
3. **Cierre bajo Uniones Numerables ($\sigma$-aditividad estructural):**
   $$\text{Dada una secuencia } \{A_i\}_{i=1}^{\infty} \subseteq \mathcal{F}, \implies \bigcup_{i=1}^{\infty} A_i \in \mathcal{F}$$


---

### 3. Medida de Probabilidad ($P$)
Una **medida de probabilidad** $P$ es una función de conjunto cuyo dominio es la $\sigma$-álgebra $\mathcal{F}$ y su codominio es el intervalo real $[0, 1]$:
$$P: \mathcal{F} \longrightarrow [0, 1] \subset \mathbb{R}$$

Para ser una medida norma de probabilidad sobre el espacio medible $(\Omega, \mathcal{F})$, $P$ debe satisfacer dos condiciones fundamentales:

1. **Normalización (Unitariedad):**
   $$P(\Omega) = 1$$
2. **$\sigma$-aditividad (Aditividad Numerable):**
   Para cualquier colección numerable de eventos disjuntos a pares (mutuamente excluyentes) $\{A_i\}_{i=1}^{\infty} \subseteq \mathcal{F}$ tales que $A_i \cap A_j = \emptyset$ para todo $i \neq j$:
   $$P\left( \bigcup_{i=1}^{\infty} A_i \right) = \sum_{i=1}^{\infty} P(A_i)$$

---  
## 2. Axiomas de Kolmogorov y Propiedades Derivadas

### Axiomas de Kolmogorov (1933):
1. **No negatividad:** Para todo $A \in \mathcal{F}$, $P(A) \geq 0$.
2. **Unitariedad:** $P(\Omega) = 1$.
3. **$\sigma$-aditividad:** Si $A_1, A_2, A_3, \dots$ es una secuencia de eventos mutuamente excluyentes ($A_i \cap A_j = \emptyset$ para $i \neq j$), entonces:
   $$P\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} P(A_i)$$

### Propiedades Derivadas:
- **Evento Imposible:** $P(\emptyset) = 0$.
- **Complemento:** $P(A^c) = 1 - P(A)$.
- **Regla de la Adición (General):** $P(A \cup B) = P(A) + P(B) - P(A \cap B)$.
- **Monotonía:** Si $A \subseteq B$, entonces $P(A) \leq P(B)$.

## 3. Probabilidad Condicional, Independencia y Teoremas Fundamentales

### 1. Probabilidad Condicional
La probabilidad condicional mide la probabilidad de que ocurra un evento $A$ sabiendo previamente que ha ocurrido un evento $B$ ($P(B) > 0$). 

* **Reducción del Espacio Muestral:** Intuitivamente, la información del evento $B$ reduce el universo de posibilidades: el espacio muestral deja de ser $\Omega$ y se convierte en $B$.
* **Fórmula:**
  $$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

---

### 2. Independencia de eventos
Dos eventos $A$ y $B$ son independientes si la ocurrencia de uno no altera la probabilidad de ocurrencia del otro (es decir, $P(A \mid B) = P(A)$).

* **Regla de Multiplicación:** Dos eventos son independientes si y solo si la probabilidad de su intersección es el producto de sus probabilidades individuales:
  $$P(A \cap B) = P(A) \cdot P(B)$$

---

### 3. Regla de la Probabilidad Total
Nos permite calcular la probabilidad general de un evento $A$ dividiéndolo a través de una partición exhaustiva y mutuamente excluyente del espacio muestral ($B_1, B_2, \dots, B_k$).

* **Fórmula:**
  $$P(A) = \sum_{i=1}^k P(A \cap B_i) = \sum_{i=1}^k P(A \mid B_i) P(B_i)$$

---

### 4. Teorema de Bayes
El Teorema de Bayes invierte el razonamiento condicional: permite actualizar la probabilidad de una hipótesis o causa $B_j$ tras haber observado cierta evidencia o evento $A$.

* **Fórmula:**
  $$P(B_j \mid A) = \frac{P(A \mid B_j) P(B_j)}{\sum_{i=1}^k P(A \mid B_i) P(B_i)}$$
* **Componentes:**
  - $P(B_j)$: Probabilidad *a priori* de la causa.
  - $P(A \mid B_j)$: Verosimilitud de la evidencia dada la causa.
  - $P(B_j \mid A)$: Probabilidad *a posteriori* corregida.

---  
## 4. Cuadro Comparativo: Enfoques de la Probabilidad

| Enfoque | Condición de Aplicabilidad | Fórmula Principal | Naturaleza del Espacio Muestral |
| :--- | :--- | :--- | :--- |
| **Clásico (Laplaciano)** | Resultados equiprobables y finitos | $P(A) = \frac{\mid A \mid}{\mid \Omega \mid} = \frac{\text{Casos Favorables}}{\text{Casos Posibles}}$ | Finito y simétrico |
| **Frecuentista** | Experimentos repetibles en condiciones idénticas | $P(A) = \lim_{N \to \infty} \frac{n_A}{N}$ | Infinito / Asintótico |
| **Geométrico** | Variable continua en subconjuntos de $\mathbb{R}^n$ | $P(A) = \frac{\text{Medida}(A)}{\text{Medida}(\Omega)}$ | Continuo (Área, Volumen, Longitud) |

## 5. Simulaciones Prácticas en Python

### Importación de librerías base

```python
import numpy as np
import matplotlib.pyplot as plt

# Fijar semilla para reproducibilidad
np.random.seed(42)
```

### Ejemplo 1: Enfoque Frecuentista - Estabilización de la Frecuencia Relativa
Estudiamos la proporción acumulada de 'Águilas' (1) al lanzar una moneda justa $N$ veces, observando la convergencia empírica hacia el valor teórico 0.5.

```python
num_lanzamientos = 10000
# Generar lanzamientos: 1 = Águila, 0 = Sol
lanzamientos = np.random.choice([0, 1], size=num_lanzamientos, p=[0.5, 0.5])

# Calcular frecuencia relativa acumulada
frecuencias_acumuladas = np.cumsum(lanzamientos) / np.arange(1, num_lanzamientos + 1)

# Graficar resultado
plt.figure(figsize=(10, 5))
plt.plot(frecuencias_acumuladas, label='Frecuencia Relativa Observada', color='navy', linewidth=1)
plt.axhline(y=0.5, color='red', linestyle='--', label='Probabilidad Teórica $P(A) = 0.5$')
plt.xlabel('Número de Lanzamientos $N$')
plt.ylabel('Frecuencia Relativa $f_A = n_A / N$')
plt.title('Estabilización Frecuentista de la Probabilidad')
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.legend()
plt.show()

print(f"Frecuencia relativa final con N={num_lanzamientos}: {frecuencias_acumuladas[-1]:.4f}")
```

Como se puede observar en la gráfica anterior, después de realizar un número considerable de simulaciones; la probabilidad de que caiga águila al largo plazo se acerca a su probabilidad teoríca $P(A) = 0.5$

### Ejemplo 2: Enfoque Clásico y Teorema de Bayes en Control de Calidad

**Planteamiento del Problema:**

Definamos a $B_n$ como el número de máquina de la fabrica, y sea D el evento de seleccionar un componente defectuoso \
Sabemos que:
- La máquina 1 ($B_1$): produce el 50% de las piezas, con una tasa de defecto de 0.20 ie $P(D \mid B_1) = 0.02$.
- La máquina 2 ($B_2$): produce el 30% de las piezas, con una tasa de defecto de 0.50 ie $P(D \mid B_2) = 0.05$.
- La máquina 3 ($B_3$): produce el 20% de las piezas, con una tasa de defecto de 0.10 ie $P(D \mid B_3) = 0.10$.

Calcularemos $P(B_3 \mid D)$ teóricamente con el Teorema de Bayes y lo verificaremos mediante conteo de un espacio muestral simulado.

```python
# Cálculo Teórico
P_B = np.array([0.50, 0.30, 0.20]) #la porporción de los lotes de componentes que produce cada máquina
P_D_dado_B = np.array([0.02, 0.05, 0.10]) # La tasa/probabilidad de que la máquina produzca un articulo defectuoso.

# Probabilidad Total P(D)
P_D = np.sum(P_D_dado_B * P_B) #Calculamos la probabilidad de que un articulo salga defectuoso independientemente de la máquina que lo fabrique

# Bayes para B3 dado D
P_B3_dado_D_teorico = (P_D_dado_B[2] * P_B[2]) / P_D  # P_D_dado_B[2] * P_B[2] Es la probabilidad de que un articulo defectuoso lo fabrique la maquina 3.

print("=== CÁLCULO TEÓRICO (Usando el teorema de Bayes)===")
print(f"Probabilidad Total de Defecto P(D): {P_D:.4f}")
print(f"Probabilidad P(B3 | D): {P_B3_dado_D_teorico:.4f}")

# Simulación de Conteo Directo (Espacio Finito)
N = 1000000

# Asignar máquinas según proporciones P(B)
maquinas = np.random.choice([1, 2, 3], size=N, p=P_B)

# Determinar estado defectuoso por cada pieza
prob_defecto_por_pieza = np.where(maquinas == 1, 0.02, np.where(maquinas == 2, 0.05, 0.10))
defectuosas = np.random.rand(N) < prob_defecto_por_pieza

# Filtrar casos favorables
piezas_defectuosas = defectuosas.sum()
defectuosas_m3 = ((maquinas == 3) & defectuosas).sum()

P_B3_dado_D_simulado = defectuosas_m3 / piezas_defectuosas

print("\n=== CÁLCULO EMPÍRICO POR CONTEO ===")
print(f"Piezas defectuosas encontradas: {piezas_defectuosas}")
print(f"Probabilidad Empírica P(B3 | D): {P_B3_dado_D_simulado:.4f}")
```

### Simulación 3: Enfoque Geométrico - El Problema del Encuentro en $\mathbb{R}^2$

**Planteamiento:** Dos personas acuerdan encontrarse entre las 12:00 y las 13:00 hrs (60 minutos). Cada una esperará un máximo de 15 minutos. ¿Cuál es la probabilidad de que se encuentren?

- Espacio Muestral: $\Omega = [0, 60] \times [0, 60] \subset \mathbb{R}^2$, con $\text{Área}(\Omega) = 60 \times 60 = 3600$.
- Condición de encuentro: $|X - Y| \leq 15$.
- Solución analítica:
  $$\text{Área}(E^c) = 2 \times \left(\frac{45 \times 45}{2}\right) = 2025$$
  $$\text{Área}(E) = 3600 - 2025 = 1575$$
  $$P(E) = \frac{1575}{3600} = \frac{7}{16} = 0.4375$$

```python
num_puntos = 100000
# Simular tiempos de llegada continuos uniformes en [0, 60]
X = np.random.uniform(0, 60, num_puntos)
Y = np.random.uniform(0, 60, num_puntos)

# Condición |X - Y| <= 15
encuentros = np.abs(X - Y) <= 15
prob_geometrica_simulada = encuentros.mean()

print(f"Probabilidad Teórica: {1575/3600:.4f}")
print(f"Probabilidad Geométrico-Simulada: {prob_geometrica_simulada:.4f}")

# Visualización de la región continua
plt.figure(figsize=(7, 7))
plt.scatter(X[::50], Y[::50], c=encuentros[::50], cmap='coolwarm', alpha=0.6, s=10, label='Submuestra de puntos')
plt.plot([0, 45], [15, 60], 'k--', label='Límite $|X - Y| = 15$')
plt.plot([15, 60], [0, 45], 'k--')
plt.xlim(0, 60)
plt.ylim(0, 60)
plt.xlabel('Tiempo de llegada Persona X (min)')
plt.ylabel('Tiempo de llegada Persona Y (min)')
plt.title('Enfoque Geométrico: Problema del Encuentro en $\mathbb{R}^2$')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```

#### Ejemplo C: Probabilidad Geométrica Unidimensional (Ross)
* **Contexto:** Un cable de longitud $L$ se rompe en un punto $X \in [0, L]$ de forma uniforme.
* **Condición:** La parte más larga mide al menos el doble de la parte más corta ($\max \geq 2 \cdot \min$).
* **Resolución por Medida de Lebesgue en $\mathbb{R}$:**
  - Evento $E = \left[0, \frac{L}{3}\right] \cup \left[\frac{2L}{3}, L\right]$
  - $P(E) = \frac{\text{Medida}(E)}{\text{Medida}(\Omega)} = \frac{\frac{L}{3} + \frac{L}{3}}{L} = \frac{2}{3}$
